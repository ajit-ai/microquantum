## 1. Setup

Working against the **installed** `microquantum` package; no `sys.path` manipulation.

`run_example` executes an example file exactly like running it from the command
line (`__name__ == '__main__'`, so `main()` blocks run), including the example's own
assertions.  A failing assertion surfaces as a cell error.

```python
import os
from pathlib import Path


def _repo_root() -> Path:
    """Locate the repo root via a known example file."""
    start = Path.cwd().resolve()
    for p in (start, *start.parents):
        if (p / "examples" / "01_basic_circuits.py").exists():
            return p
    return start


_ROOT = _repo_root()


def run_example(rel: str) -> None:
    """Run an official example script as the main module.

    Mirrors ``uv run python examples/<rel>`` from the repo root:
    cwd is the repo root, ``__name__ == '__main__'`` so main() runs,
    and the example's own assertions are executed.
    """
    full = (_ROOT / rel).resolve()
    source = full.read_text(encoding="utf-8")
    namespace = {"__name__": "__main__", "__file__": str(full)}
    print(f"\n===== {rel} =====")
    prev = Path.cwd()
    os.chdir(_ROOT)
    try:
        exec(compile(source, str(full), "exec"), namespace)
    finally:
        os.chdir(prev)
```
